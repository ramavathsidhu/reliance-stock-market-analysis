```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.plot(df['Close'])

plt.title('Reliance Industries Closing Price')
plt.xlabel('Trading Days')
plt.ylabel('Stock Price (INR)')

plt.show()
```

```python
df['MA50']=df['Close'].rolling(window=50).mean()
df['MA200']=df['Close'].rolling(window=200).mean()
df[['Close','MA50','MA200']].tail()
```

```python
plt.figure(figsize=(12,5))

plt.plot(df['Close'], label='Closing Price')
plt.plot(df['MA50'], label='50-Day MA')
plt.plot(df['MA200'], label='200-Day MA')

plt.title('Reliance Industries Moving Average')
plt.xlabel('Date')
plt.ylabel('Stock Price (INR)')

plt.legend()
plt.show()
```

```python
df['Daily_Return']=df['Close'].pct_change()
df[['Close','Daily_Return']].tail(10)
```

```python
average_return=df['Daily_Return'].mean()*100
print("Average Daily Return:",average_return,"%")
```

```python
daily_volatility=df['Daily_Return'].std()*100
print("Daily Volatility:",daily_volatility,"%")
```

```python
import numpy as np
annual_volatility=df['Daily_Return'].std()*np.sqrt(252)*100
print("Daily Volatility:",annual_volatility,"%")
```

```python
df['Daily_Return'].describe()
```

```python
daily_volatility=df['Daily_Return'].dropna().std()*100
print("Daily Volatility:",daily_volatility,"%")
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['Daily_Return'])
plt.title('reliance industries Daily Returns')
plt.ylabel('Daily Return')
plt.xlabel('trading days')
plt.show()
```

```python
best_day = df['Daily_Return'].max()

worst_day = df['Daily_Return'].min()

print("Best Daily Return:", best_day * 100, "%")

print("Worst Daily Return:", worst_day * 100, "%")
```

```python
print("Best Day:",df['Daily_Return'].idxmax())
print("Worst Day:",df['Daily_Return'].idxmin())
```

```python
df['Cumulative_Return'] = (1 + df['Daily_Return']).cumprod() - 1

print(df[['Close', 'Cumulative_Return']].tail())
```

```python
plt.figure(figsize=(12,5))

plt.plot(df['Cumulative_Return'] * 100)

plt.title('Reliance Industries Cumulative Returns')
plt.xlabel('Trading Days')
plt.ylabel('Cumulative Return (%)')

plt.axhline(y=0, linestyle='--')

plt.show()
```

```python
monthly_returns = df['Close'].resample('ME').last().pct_change() * 100

print(monthly_returns.tail(12))
```

```python
import matplotlib.pyplot as plt

monthly_returns.plot(figsize=(12,5))

plt.title('Reliance Industries Monthly Stock Returns')
plt.xlabel('Month')
plt.ylabel('Monthly Returns (%)')
plt.grid()
plt.show()
```

```python
yearly_returns = df['Close'].resample('YE').last().pct_change() * 100

print(yearly_returns)
```

```python
plt.figure(figsize=(12,5))

yearly_returns.plot(kind='bar')

plt.title('Reliance Industries Yearly Returns')
plt.xlabel('Year')
plt.ylabel('Yearly Return (%)')

plt.axhline(y=0, linestyle='--')

plt.show()
```

```python
start_price = df['Close'].iloc[0]
end_price = df['Close'].iloc[-1]

total_return = ((end_price / start_price) - 1) * 100

print("Starting Price:", start_price)
print("Ending Price:", end_price)
print("Total Return:", total_return, "%")
```

```python
years = (df.index[-1] - df.index[0]).days / 365.25

cagr = ((end_price / start_price) ** (1 / years) - 1) * 100

print("CAGR:", cagr, "%")
```

```python
print("RELIANCE INDUSTRIES STOCK ANALYSIS")
print("----------------------------------")

print("Starting Price:", start_price)
print("Ending Price:", end_price)
print("Total Return:", round(total_return, 2), "%")
print("CAGR:", round(cagr, 2), "%")
print("Daily Volatility:", round(daily_volatility, 2), "%")
print("Annual Volatility:", round(annual_volatility, 2), "%")
print("Best Daily Return:", round(best_day * 100, 2), "%")
print("Worst Daily Return:", round(worst_day * 100, 2), "%")
```

```python
close = df['Close'].iloc[:, 0]

daily_returns = close.pct_change()

start_price = float(close.iloc[0])
end_price = float(close.iloc[-1])

total_return = ((end_price / start_price) - 1) * 100

years = (close.index[-1] - close.index[0]).days / 365.25

cagr = ((end_price / start_price) ** (1 / years) - 1) * 100

daily_volatility = daily_returns.std() * 100

annual_volatility = daily_returns.std() * (252 ** 0.5) * 100

best_day = daily_returns.max()
worst_day = daily_returns.min()

print("RELIANCE INDUSTRIES STOCK ANALYSIS")
print("-----------------------------------")
print("Starting Price:", round(start_price, 2))
print("Ending Price:", round(end_price, 2))
print("Total Return:", round(total_return, 2), "%")
print("CAGR:", round(cagr, 2), "%")
print("Daily Volatility:", round(daily_volatility, 2), "%")
print("Annual Volatility:", round(annual_volatility, 2), "%")
print("Best Daily Return:", round(best_day * 100, 2), "%")
print("Worst Daily Return:", round(worst_day * 100, 2), "%")
```
